```python
from typing import Literal
from typing_extensions import TypedDict
from langchain_openai import ChatOpenAI

class Router(TypedDict):
    logic: str

    type: Literal[
        "general-query",
        "additional-query",
        "graphrag-query",
        "image-query",
        "file-query"
    ]

    question: str


model = ChatOpenAI(
    model="gpt-4o"
)

router_model = model.with_structured_output(
    Router
)

result = router_model.invoke(
"""
Classify the user query.

User Query:
Do you have smart locks?
"""
)

print(result)
print(result["type"])
```

{'logic': 'The user query is asking about the availability or existence of smart locks, likely inquiring about whether a certain company, service, or platform offers smart locks for purchase or use.', 'type': 'general-query', 'question': 'Do you have smart locks?'}

## LCEL -> LangChain Language

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_openai import ChatOpenAI

prompt = ChatPromptTemplate.from_template(
    "Tell me about {topic}"
)

model = ChatOpenAI(
    model="gpt-4o"
)
#|->pipeline Runnable Chain
chain = prompt | model

result = chain.invoke(
    {"topic": "AI Agent"}
)

print(result.content)
```

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_openai import ChatOpenAI
from langchain_core.output_parsers import StrOutputParser

# Prompt
prompt = ChatPromptTemplate.from_template(
    "Please introduce {topic} in 50 words."
)

# Model
model = ChatOpenAI(
    model="gpt-4o"
)

# Parser
parser = StrOutputParser()

# LCEL Chain
chain = prompt | model | parser

# Invoke
result = chain.invoke(
    {"topic": "Artificial Intelligence"}
)

print(result)
```

```python
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import JsonOutputParser
from langchain_openai import ChatOpenAI

model = ChatOpenAI(
    model="gpt-4o"
)

parser = JsonOutputParser()

prompt = PromptTemplate(
    template="""
Return information about the book in JSON format.

Book: {book}

Example:
{{
    "title": "...",
    "author": "...",
    "year": ...
}}
"""
)

chain = prompt | model | parser

result = chain.invoke(
    {"book": "Harry Potter"}
)

print(result)
print(type(result))
```

```python
# Retriever
from langchain_core.documents import Document

docs = [
    Document(
        page_content="Apple Home Smart Lock Max supports fingerprint unlock."
    ),
    Document(
        page_content="Google Nest Camera provides 2K video monitoring."
    ),
    Document(
        page_content="Amazon Echo Hub is a smart home control center."
    )
]

```

```python
from langchain_core.vectorstores import InMemoryVectorStore
from langchain_openai import OpenAIEmbeddings

vectorstore = InMemoryVectorStore.from_documents(
    docs,
    OpenAIEmbeddings()
)

retriever = vectorstore.as_retriever()
```

```python
question = "Do you have smart locks?"

results = retriever.invoke(question)

for doc in results:
    print(doc.page_content)
```

```python
retriever = vectorstore.as_retriever(
    search_kwargs={
        "k": 1
    }
)

results = retriever.invoke(
    "Do you have smart locks?"
)

for i, doc in enumerate(results, start=1):
    print(f"Document {i}:")
    print(doc.page_content)
    print("-" * 50)
```
